## 📦 Cell 1 — Install Dependencies
> Run once. Safe to skip if already installed.

````python
import subprocess, sys

packages = [
    'earthengine-api', 'geemap', 'geopandas', 'rasterio',
    'numpy', 'pandas', 'scikit-learn', 'matplotlib', 'seaborn',
    'folium', 'streamlit', 'streamlit-folium',
    'langchain', 'langchain-openai', 'langgraph==0.0.38',
    'requests', 'python-dotenv', 'shapely', 'plotly', 'Pillow', 'ipywidgets'
]

print('📦 Installing packages...\n')
failed = []
for pkg in packages:
    result = subprocess.run(
        [sys.executable, '-m', 'pip', 'install', pkg, '-q'],
        capture_output=True, text=True
    )
    status = '✅' if result.returncode == 0 else '❌'
    if result.returncode != 0:
        failed.append(pkg)
    print(f'  {status} {pkg}')

print('\n' + '='*45)
if failed:
    print(f'❌ Failed packages: {failed}')
    print('   Try: conda install -c conda-forge rasterio geopandas -y')
else:
    print('🎉 All packages installed successfully!')
    print('   ➡ Restart kernel if this is your first install (Kernel → Restart)')
````

## 📡 Cell 4 — Load / Generate Field Data

````python
from src.soc_model import generate_synthetic_training_data

DISTRICT = 'Wardha'    # Options: Wardha, Yavatmal, Amravati, Nashik
N_SAMPLES = 500        # Number of field pixel samples

if USE_REAL_GEE:
    # ── Real GEE pipeline ──────────────────────────────────────────
    from src.ingestion import (
        initialize_gee, get_study_area,
        get_sentinel2_collection, get_soilgrids_soc, sample_pixels
    )
    initialize_gee()
    geometry  = get_study_area(DISTRICT)
    composite = get_sentinel2_collection(geometry,
                    start_date='2023-10-01', end_date='2024-03-31')
    soc_layer = get_soilgrids_soc(geometry)
    raw_df    = sample_pixels(composite, soc_layer, geometry, n_samples=N_SAMPLES)
    raw_df.to_csv('outputs/raw_gee_samples.csv', index=False)
    print(f'💾 GEE data saved to outputs/raw_gee_samples.csv')

else:
    # ── Synthetic data ─────────────────────────────────────────────
    raw_df = generate_synthetic_training_data(N_SAMPLES)
    raw_df.to_csv('outputs/raw_samples.csv', index=False)
    print(f'💾 Synthetic data saved to outputs/raw_samples.csv')

print(f'\n📊 Dataset shape   : {raw_df.shape}')
print(f'📊 Columns         : {list(raw_df.columns)}')
print(f'📊 Null values     : {raw_df.isnull().sum().sum()}')
print()
raw_df.head()
````

## 📐 Cell 5 — Compute Spectral Indices

| Index | Formula | Purpose |
|-------|---------|--------|
| NDVI  | (NIR-Red)/(NIR+Red) | Vegetation health |
| BSI   | ((SWIR1+Red)-(NIR+Blue))/(...) | Bare soil exposure |
| NDTI  | (SWIR1-SWIR2)/(SWIR1+SWIR2) | Tillage intensity |
| NDWI  | (Green-NIR)/(Green+NIR) | Water content |
| EVI   | 2.5*(NIR-Red)/(NIR+6R-7.5B+1) | Enhanced vegetation |
| CRC   | (SWIR1-Green)/(SWIR1+Green) | Crop residue cover |

````python
from src.indices import compute_indices_df

df = compute_indices_df(raw_df)

# Show index statistics
index_cols = ['NDVI', 'BSI', 'NDTI', 'NDWI', 'EVI', 'CRC']
print('\n📐 Spectral Index Statistics:')
print('='*55)
stats = df[index_cols].describe().round(3)
print(stats.to_string())
print('\n✅ All 6 spectral indices computed successfully')
````

## 🌾 Cell 6 — Classify Land Practices

````python
from src.indices import add_practice_labels

df = add_practice_labels(df)

print('\n🌾 Land Practice Distribution:')
print('='*45)
counts = df['LandPractice'].value_counts()
total  = len(df)
for practice, count in counts.items():
    bar = '█' * int(count / total * 30)
    pct = round(count / total * 100, 1)
    print(f'  {pct:5.1f}%  {bar:<30}  {practice}')
print(f'\n  Total fields: {total}')
````

## 🤖 Cell 7 — Train ML Models
> Trains Random Forest for SOC regression + land practice classification

````python
from src.soc_model import SOCPredictor, PracticeClassifier

print('Training SOC Estimator (Random Forest Regressor)...')
print('-'*50)
soc_model = SOCPredictor(n_estimators=200)
soc_metrics = soc_model.train(df)

print('\nTraining Practice Classifier (Random Forest Classifier)...')
print('-'*50)
prac_clf = PracticeClassifier()
clf_metrics = prac_clf.train(df)

print('\n' + '='*50)
print('📊 FINAL MODEL SUMMARY')
print('='*50)
print(f"  SOC Estimator  →  R² = {soc_metrics['R2']}  |  RMSE = {soc_metrics['RMSE']} g/kg")
print(f"  CV R²          →  {soc_metrics['CV_R2_mean']} ± {soc_metrics['CV_R2_std']}")
print(f"  Classifier     →  Accuracy = {round(clf_metrics['accuracy']*100, 1)}%")
print('='*50)

# Feature importances
print('\n🔍 Top 5 Most Important Features for SOC:')
importances = sorted(soc_model.feature_importances_.items(), key=lambda x: x[1], reverse=True)
for feat, imp in importances[:5]:
    bar = '█' * int(imp * 100)
    print(f'  {feat:<6}  {imp:.4f}  {bar}')

# Save models
soc_model.save('outputs/soc_model.pkl')
print('\n💾 Models saved to outputs/')
````

## 🔮 Cell 8 — Run Predictions on Field Samples

````python
from src.soc_model import generate_synthetic_training_data
from src.indices import compute_indices_df, add_practice_labels, get_carbon_potential

# Generate a fresh test set (unseen by model)
test_raw = generate_synthetic_training_data(150)
test_df  = compute_indices_df(test_raw)
test_df  = add_practice_labels(test_df)

# Run predictions
test_df = soc_model.predict(test_df)
test_df = prac_clf.predict_proba_df(test_df)

# Add carbon potential
test_df['CreditPotential_INR'] = test_df['PredictedPractice'].apply(
    lambda p: get_carbon_potential(p)['credit_potential_per_ha']
)
test_df['Recommendation'] = test_df['PredictedPractice'].apply(
    lambda p: get_carbon_potential(p)['recommendation']
)

# Save predictions
test_df.to_csv('outputs/field_predictions.csv', index=False)

print('✅ Predictions complete!')
print(f'   Fields predicted : {len(test_df)}')
print(f'   Avg SOC          : {test_df["SOC_predicted"].mean():.2f} g/kg')
print(f'   Healthy fields   : {(test_df["SOC_predicted"] >= 15).sum()} ({round((test_df["SOC_predicted"] >= 15).mean()*100, 1)}%)')
print(f'   Degraded fields  : {(test_df["SOC_predicted"] < 8).sum()} ({round((test_df["SOC_predicted"] < 8).mean()*100, 1)}%)')
print()
display(test_df[['SOC_predicted','SOC_category','PredictedPractice','PracticeConfidence','CreditPotential_INR']].head(10))
````

## 📊 Cell 9 — Visualizations

````python
fig, axes = plt.subplots(2, 2, figsize=(16, 12))
fig.patch.set_facecolor('#0e1117')
fig.suptitle('🌱 Soil Carbon Analysis — Wardha, Maharashtra', 
             fontsize=16, color='white', fontweight='bold', y=0.98)

TEAL  = '#4ecdc4'
RED   = '#ff6b6b'
GOLD  = '#ffd93d'
GREEN = '#6bcb77'

# ── Plot 1: SOC Distribution ──────────────────────────────────────
ax1 = axes[0, 0]
ax1.set_facecolor('#1a1a2e')
ax1.hist(test_df['SOC_predicted'], bins=25, color=TEAL, edgecolor='#0e1117', alpha=0.85)
ax1.axvline(20,  color=RED,  linestyle='--', linewidth=2, label='Healthy threshold (20 g/kg)')
ax1.axvline(test_df['SOC_predicted'].mean(), color=GOLD,
            linestyle=':', linewidth=2, label=f'Mean: {test_df["SOC_predicted"].mean():.1f} g/kg')
ax1.set_title('SOC Distribution', color='white', fontsize=13, pad=10)
ax1.set_xlabel('SOC (g/kg)', color='#aaa')
ax1.set_ylabel('Field Count', color='#aaa')
ax1.tick_params(colors='#888')
ax1.legend(fontsize=9)
ax1.spines['bottom'].set_color('#333')
ax1.spines['left'].set_color('#333')

# ── Plot 2: Practice Distribution ────────────────────────────────
ax2 = axes[0, 1]
ax2.set_facecolor('#1a1a2e')
counts  = test_df['PredictedPractice'].value_counts()
colors  = [TEAL, GREEN, GOLD, RED, '#a78bfa', '#f9844a']
wedges, texts, autotexts = ax2.pie(
    counts.values,
    labels=None,
    autopct='%1.1f%%',
    colors=colors[:len(counts)],
    startangle=90,
    pctdistance=0.82,
    wedgeprops={'edgecolor': '#0e1117', 'linewidth': 2}
)
for t in autotexts: t.set_color('white'); t.set_fontsize(9)
ax2.legend(wedges, counts.index, loc='lower center',
           bbox_to_anchor=(0.5, -0.25), fontsize=8, ncol=2,
           framealpha=0.1, labelcolor='white')
ax2.set_title('Land Practice Distribution', color='white', fontsize=13, pad=10)

# ── Plot 3: SOC by Practice Box Plot ─────────────────────────────
ax3 = axes[1, 0]
ax3.set_facecolor('#1a1a2e')
practices   = test_df['PredictedPractice'].unique()
soc_by_prac = [test_df[test_df['PredictedPractice'] == p]['SOC_predicted'].values
               for p in practices]
bp = ax3.boxplot(soc_by_prac, patch_artist=True, notch=False,
                 medianprops={'color': 'white', 'linewidth': 2})
for patch, color in zip(bp['boxes'], colors[:len(practices)]):
    patch.set_facecolor(color); patch.set_alpha(0.75)
for element in ['whiskers', 'caps', 'fliers']:
    for item in bp[element]: item.set_color('#555')
ax3.axhline(20, color=RED, linestyle='--', linewidth=1.5, label='Healthy threshold')
ax3.set_xticks(range(1, len(practices)+1))
ax3.set_xticklabels([p[:18] for p in practices], rotation=25, ha='right', fontsize=8, color='#aaa')
ax3.set_title('SOC by Land Practice', color='white', fontsize=13, pad=10)
ax3.set_ylabel('SOC (g/kg)', color='#aaa')
ax3.tick_params(colors='#888')
ax3.legend(fontsize=9)
ax3.spines['bottom'].set_color('#333'); ax3.spines['left'].set_color('#333')

# ── Plot 4: Carbon Credit Potential ──────────────────────────────
ax4 = axes[1, 1]
ax4.set_facecolor('#1a1a2e')
credit_by_prac = test_df.groupby('PredictedPractice')['CreditPotential_INR'].mean().sort_values(ascending=True)
bars = ax4.barh(credit_by_prac.index, credit_by_prac.values,
                color=colors[:len(credit_by_prac)], edgecolor='#0e1117', alpha=0.85)
for bar, val in zip(bars, credit_by_prac.values):
    ax4.text(val + 50, bar.get_y() + bar.get_height()/2,
             f'₹{val:,.0f}', va='center', color='white', fontsize=9)
ax4.set_title('Carbon Credit Potential (₹/ha/year)', color='white', fontsize=13, pad=10)
ax4.set_xlabel('₹ INR per ha per year', color='#aaa')
ax4.tick_params(colors='#888')
ax4.set_yticklabels([p[:22] for p in credit_by_prac.index], fontsize=8, color='#aaa')
ax4.spines['bottom'].set_color('#333'); ax4.spines['left'].set_color('#333')

plt.tight_layout()
plt.savefig('outputs/analysis_charts.png', dpi=150,
            bbox_inches='tight', facecolor='#0e1117')
plt.show()
print('✅ Charts saved to outputs/analysis_charts.png')
````

## 🗺️ Cell 10 — Interactive Folium Map
> Click any dot on the map to see field-level SOC + practice details

````python
import folium

# Assign coordinates if not present
if 'latitude' not in test_df.columns:
    np.random.seed(42)
    test_df['latitude']  = np.random.uniform(20.40, 21.10, len(test_df))
    test_df['longitude'] = np.random.uniform(78.35, 79.10, len(test_df))

def soc_color(soc):
    if soc < 5:    return '#d73027'
    elif soc < 8:  return '#fc8d59'
    elif soc < 12: return '#fee090'
    elif soc < 16: return '#91cf60'
    elif soc < 22: return '#1a9850'
    else:          return '#006837'

center_lat = test_df['latitude'].mean()
center_lon = test_df['longitude'].mean()

m = folium.Map(
    location=[center_lat, center_lon],
    zoom_start=10,
    tiles='CartoDB dark_matter'
)

for _, row in test_df.iterrows():
    color  = soc_color(row['SOC_predicted'])
    radius = max(5, row.get('PracticeConfidence', 70) / 15)
    popup  = f"""
    <div style='font-family:monospace;width:230px;font-size:12px;'>
    <b>📍 Field Report</b><hr>
    <b>SOC:</b>       {row['SOC_predicted']} g/kg<br>
    <b>Category:</b>  {row['SOC_category']}<br>
    <b>Practice:</b>  {row['PredictedPractice']}<br>
    <b>Confidence:</b>{row.get('PracticeConfidence','N/A')}%<br>
    <b>NDVI:</b>      {round(row.get('NDVI',0), 3)}<br>
    <b>BSI:</b>       {round(row.get('BSI',0), 3)}<br><hr>
    <b>💰 Credit:</b> ₹{row['CreditPotential_INR']:,}/ha/yr<br>
    <i>{row['Recommendation'][:80]}...</i>
    </div>
    """
    folium.CircleMarker(
        location=[row['latitude'], row['longitude']],
        radius=radius,
        color=color, fill=True, fill_color=color, fill_opacity=0.75,
        popup=folium.Popup(popup, max_width=260),
        tooltip=f"SOC: {row['SOC_predicted']} g/kg | {row['PredictedPractice'][:20]}"
    ).add_to(m)

legend_html = '''
<div style="position:fixed;bottom:30px;left:30px;z-index:9999;
            background:rgba(0,0,0,0.85);padding:14px;border-radius:10px;
            font-family:monospace;color:white;font-size:12px;border:1px solid #444;">
<b>🌱 SOC Level (g/kg)</b><br><br>
<span style="color:#d73027">●</span> &lt; 5  — Critical<br>
<span style="color:#fc8d59">●</span> 5–8  — Very Low<br>
<span style="color:#fee090">●</span> 8–12 — Low<br>
<span style="color:#91cf60">●</span> 12–16 — Moderate<br>
<span style="color:#1a9850">●</span> 16–22 — Good<br>
<span style="color:#006837">●</span> &gt; 22 — Excellent
</div>'''
m.get_root().html.add_child(folium.Element(legend_html))

m.save('outputs/soc_map.html')
print('✅ Map saved to outputs/soc_map.html')
print('   Open this file in your browser for full interactivity')
print('   Or it will display below:')
m
````

## 🤖 Cell 11 — Run the 5-Node Agentic Pipeline

This is the **core** of the project — watch each agent node reason step-by-step:

```
🔍 Perception → 📊 Analysis → 🩺 Diagnosis → 💡 Recommendation → 📋 Report
```

````python
from src.agent import run_mock_agent

# Convert predictions DataFrame to list of field dicts for agent
field_data = test_df.to_dict('records')

print('🚀 Launching Soil Carbon Agentic Pipeline...')
print('   This runs 5 reasoning nodes in sequence')
print()

# Run the full agent pipeline
result = run_mock_agent(
    field_data=field_data,
    region_name='Wardha, Maharashtra'
)

print()
print('='*55)
print('📋 AGENT REASONING TRACE')
print('='*55)
for i, step in enumerate(result['reasoning_trace']):
    node_names = ['🔍 PERCEPTION','📊 ANALYSIS','🩺 DIAGNOSIS','💡 RECOMMENDATIONS','📋 REPORT']
    print(f'\n── Node {i+1}: {node_names[i]} ──')
    if i == 3:  # Recommendations — print structured
        for rec in result['recommendations']:
            print(f"  [{rec['priority']}] {rec['practice']}")
            print(f"         Carbon: {rec['carbon_benefit']} | ₹{rec['credit_potential_inr_per_ha']:,}/ha")
    else:
        content = step.split(':', 1)[-1].strip() if ':' in step else step
        print(content[:600])
````

## 📋 Cell 12 — Print & Save Final Report

````python
from IPython.display import Markdown, display

# Display as rendered markdown in notebook
display(Markdown(result['final_report']))

# Save report files
report_path = 'outputs/soil_carbon_report_wardha.md'
with open(report_path, 'w', encoding='utf-8') as f:
    f.write(result['final_report'])

# Save recommendations as CSV
recs_df = pd.DataFrame(result['recommendations'])
recs_df.to_csv('outputs/recommendations.csv', index=False)

# Save full predictions
test_df.to_csv('outputs/field_predictions.csv', index=False)

print('\n' + '='*45)
print('💾 All outputs saved:')
print(f'   📄 outputs/soil_carbon_report_wardha.md')
print(f'   📊 outputs/field_predictions.csv')
print(f'   💡 outputs/recommendations.csv')
print(f'   🗺️  outputs/soc_map.html')
print(f'   📈 outputs/analysis_charts.png')
print('='*45)
print('\n🎉 Pipeline complete! Ready for LinkedIn demo!')
````
